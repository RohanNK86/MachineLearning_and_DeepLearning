**Here we will create a simple AI Application using the API Keys**

```python
import os
import openai
```

```python
openai_api_keys = os.environ['OPENAI_API_KEYS'] #Load the openai api keys or use the alternative open router platform to get free api keys
```

```python
llm_model = "gpt-3.5-turbo" #Load the LLM Model or any openai model that is available
```

Let's start with a direct API calls to OpenAI.

```python
def get_completion(prompt, model=llm_model):
    messages = [{"role" : "user", "content" : prompt}]
    response = openai.ChatCompletion(
        model=model,
        messages=messages,
        temperature=0.0
    )
    return response.choices[0].message['content']
```

```python
get_completion("What is 1 + 1 ?")
```

**Output** : '1 + 1 equals to 2'

This above was a simple AI Application that can answer some of the questions

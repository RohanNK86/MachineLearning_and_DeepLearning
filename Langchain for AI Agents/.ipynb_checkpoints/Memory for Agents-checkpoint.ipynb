**AI Applications don't usally remember all the past conversation we did so far, so everytime we need to give them the entire context as a content.**

**To Resolve this we give them a memory, there are different types of memory**

**LangChain: Memory Outline**

* ConversationBufferMemory
* ConversationBufferWindowMemory
* ConversationTokenBufferMemory
* ConversationSummaryMemory

**1st ConversationBufferMemory**

```python
import os
```

```python
llm = ChatOpenAI(temperature=0.0, model=llm_model)
memory = CoversationalBufferMemory()
conversation = ConversationalChain(
    llm=llm,
    memory=memory,
    verbose=False
)
```

**2nd ConversationBufferWindowMemory**

```python
from langchain.memory import ConversationalBufferWindowMemory #In this memory system this will store and retrieve the last kth interactions over time.
```

```python
memory = ConversationalBufferWindowMemory(k=1)
```

```python
llm = ChatOpenAI(temperatute=0.0,  model=llm_model)
conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=False
)
```

```python
conversation.predict(input="Hi, my name is Andrew")
```

**Output** : "Hello Andrew! It's nice to meet you. How can I assist you today?"

```python
conversation.predict(input="What is 1+1?")
```

**Output** : '1+1 equals 2. Is there anything else you would like to know?'

```python
conversation.predict(input="What is my name?")
```

**Output** : I'm sorry, I do not have access to personal information such as your name. Is there anything else you would like to know?"

Here in the buffer window memory it will remebers the past kth interations.

**3rd ConversationTokenBufferMemory**

```python
from langchain.memory import ConversationTokenBufferMemory #This method keeps the buffer of recent interactions in memory and uses the token length instead of the number of interactions.
from langchain.llms import OpenAI
```

```python
memory = ConversationTokenBufferMemory(llm=llm, max_token_limit=50)
```

```python
llm = ChatOpenAI(temperature=0.0, model=llm_model)
```

```python
from langchain.chains import ConversationChain
```

```python
conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=False
)
```

```python
memory.save_context({"input": "AI is what?!"},
                    {"output": "Amazing!"})
memory.save_context({"input": "Backpropagation is what?"},
                    {"output": "Beautiful!"})
memory.save_context({"input": "Chatbots are what?"}, 
                    {"output": "Charming!"})
```

```python
memory.load_memory_variables({})
```

**Output** : {'history': 'AI: Amazing!\nHuman: Backpropagation is what?\nAI: Beautiful!\nHuman: Chatbots are what?\nAI: Charming!'}

**4th ConversationSummaryBufferMemory**

```python
from langchain.memory import CoversationSummaryBufferMemory #This memory keeps the summary of the conversation we did over the time.
from langchain.llms import OpenAI
from langchain.chains import CoversationChain
```

```python
llm = ChatOpenAI(temperature=0.0, model=llm)
memory = CoversationSummaryBufferMemory(llm=llm, max_token_limit=100)
```

```python
conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=False
)
```

```python
# create a long, This is the context for the LLM we are giving
schedule = "There is a meeting at 8am with your product team. \
You will need your powerpoint presentation prepared. \
9am-12pm have time to work on your LangChain \
project which will go quickly because Langchain is such a powerful tool. \
At Noon, lunch at the italian resturant with a customer who is driving \
from over an hour away to meet you to understand the latest in AI. \
Be sure to bring your laptop to show the latest LLM demo."

memory = ConversationSummaryBufferMemory(llm=llm, max_token_limit=100)
memory.save_context({"input": "Hello"}, {"output": "What's up"})
memory.save_context({"input": "Not much, just hanging"},
                    {"output": "Cool"})
memory.save_context({"input": "What is on the schedule today?"}, 
                    {"output": f"{schedule}"})
```

```python
memory.load_memory_variables({})
```

**Output** : {'history': 'System: The human and AI exchange greetings and discuss the schedule for the day, including a meeting with the product team, work on the LangChain project, and a lunch meeting with a customer interested in AI. The AI provides details on each event and emphasizes the power of LangChain as a tool.'}

```python
conversation.predict(input="What would be a good demo to show?") #This will return the response only within the 100 tokens
```

**Output** : 'For the lunch meeting with the customer interested in AI, a demo focusing on the AI capabilities of LangChain would be impressive. You could showcase how LangChain uses machine learning algorithms to continuously improve translation accuracy, adapt to different languages and dialects, and provide personalized recommendations based on user preferences. Highlighting the seamless integration of AI technology into the translation process and the potential for customization based on specific industry needs would be compelling for the customer.'
